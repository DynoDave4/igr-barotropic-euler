```julia
using Pkg
using CSV
using DataFrames
using Plots
```

```julia
function read_mfem_nodes(filename)
    coords = Float64[]
    in_nodes = false
    skip = 0

    for line in eachline(filename)
        if strip(line) == "nodes"
            in_nodes = true
            skip = 4  # skip FE space metadata
            continue
        end

        if in_nodes
            if skip > 0
                skip -= 1
                continue
            end
            try
                push!(coords, parse(Float64, strip(line)))
            catch
            end
        end
    end
    return coords
end
```

```julia
function read_other(filename)
    coords = Float64[]
    in_nodes = false
    skip = 4

    for line in eachline(filename)
        if skip > 0
            skip -= 1
            continue
        end
        try
            push!(coords, parse(Float64, strip(line)))
        catch
        end
    end
    return coords
end
```

```julia
x = read_mfem_nodes("Laghos_1_mesh")
rho = read_other("Laghos_1_rho")
v = read_other("Laghos_1_v")
e = read_other("Laghos_1_e")
igrp = read_other("Laghos_1_igr");
```

```julia
pv = sortperm(x)
x = x[pv]
rho = rho[pv]
v = v[pv]
e = e[pv]
igrp = igrp[pv];
```
